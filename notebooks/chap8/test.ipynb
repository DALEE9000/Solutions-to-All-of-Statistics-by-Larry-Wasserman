```python

B = 1000

y = np.random.normal(0,1,50)
x = np.exp(y)
thetahat = skew(x)


def bootstrap():
    bootstrapsamples = []
    bootstrapsamples.append(skew(choices(x, k=len(y))))
    return bootstrapsamples;

#arg_iterable = [i for i in range(0,B)]

#with multiprocess.Pool(processes=6) as pool:
#    result = pool.map(bootstrap,arg_iterable)

if __name__ == '__main__':
    count = B
    async_results = []
    with multiprocess.Pool(processes=4) as pool:
        for _ in range(count):
            async_results.append(pool.apply_async(bootstrap))
        pool.close()
        pool.join()

    results = np.zeros([count, 2])
    for i, async_result in enumerate(async_results):
        results[i] = async_result.get()
    print(results)




sampleboot = results

seboot = np.std(sampleboot)

# alpha = 0.05
normci = (thetahat - 1.96*seboot, thetahat + 1.96*seboot)
pivotalci = (2*thetahat - np.quantile(sampleboot, 0.975), 2*thetahat - np.quantile(sampleboot, 0.025))
percentileci = (np.quantile(sampleboot, 0.025), np.quantile(sampleboot, 0.975))

print("%.4f" % thetahat, "%.4f" % seboot, "(%.4f,%.4f)" % normci, "(%.4f,%.4f)" % pivotalci, "(%.4f,%.4f)" % percentileci)
```
